```python
import pandas as pd
import os
import glob


# ==========================================
# 1. Folder containing the original CSVs
# ==========================================

input_folder = r"D:\4-1\NLP_Lab\NLP-Based-Research-Paper-Intelligence-System\data\raw_data"


# ==========================================
# 2. Output file
# ==========================================

output_file = r"D:\4-1\NLP_Lab\NLP-Based-Research-Paper-Intelligence-System\data\processed_data\merged_research_papers.csv"


# ==========================================
# 3. Required columns
# ==========================================

required_columns = [
    "Title",
    "Author",
    "Citation count",
    "Concept",
    "Domain",
    "Field",
    "Keyword",
    "Topic",
    "Topic IDs",
    "Institution",
    "Cited by",
    "Cites",
    "DOI",
    "Abstract",
    "Top 1% cited",
    "Top 10% cited"
]


# ==========================================
# 4. File name → Category mapping
# ==========================================

category_mapping = {
    "hpc.csv": "HPC",
    "iot.csv": "IOT",
    "networks.csv": "NETWORKS",
    "nlp.csv": "NLP",
    "security.csv": "SECURITY",
    "vision.csv": "VISION"
}


# ==========================================
# 5. Find CSV files
# ==========================================

csv_files = glob.glob(
    os.path.join(input_folder, "*.csv")
)

print("CSV files found:", len(csv_files))


# ==========================================
# 6. Process each CSV
# ==========================================

dataframes = []

for file in csv_files:

    filename = os.path.basename(file)

    print("\nProcessing:", filename)

    # Check whether filename has a category
    if filename not in category_mapping:
        print("Skipping:", filename)
        print("No category mapping found.")
        continue

    category = category_mapping[filename]

    # Read CSV
    df = pd.read_csv(
        file,
        encoding="utf-8-sig"
    )

    # Add category
    df["Category"] = category

    # Keep required columns + Category
    df = df[required_columns + ["Category"]]

    dataframes.append(df)

    print("Category:", category)
    print("Papers:", len(df))


# ==========================================
# 7. Merge datasets
# ==========================================

merged_df = pd.concat(
    dataframes,
    ignore_index=True
)


# ==========================================
# 8. Remove duplicate papers
# ==========================================

before = len(merged_df)

merged_df = merged_df.drop_duplicates(
    subset=["Title"],
    keep="first"
)

after = len(merged_df)

print("\nDuplicates removed:", before - after)


# ==========================================
# 9. Save
# ==========================================

merged_df.to_csv(
    output_file,
    index=False,
    encoding="utf-8-sig"
)


# ==========================================
# 10. Show result
# ==========================================

print("\n===================================")
print("MERGING COMPLETED")
print("===================================")

print("Total papers:", len(merged_df))

print("\nPapers per category:")
print(merged_df["Category"].value_counts())

print("\nSaved to:")
print(output_file)
```
