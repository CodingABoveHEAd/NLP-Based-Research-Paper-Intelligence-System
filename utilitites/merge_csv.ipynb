```python
import pandas as pd
import os
import glob


# ==========================================
# 1. Folder containing the CSV files
# ==========================================

input_folder = r"D:\4-1\NLP_Lab\NLP-Based-Research-Paper-Intelligence-System\data\raw_data"


# ==========================================
# 2. Output folder
# ==========================================

output_folder = r"D:\4-1\NLP_Lab\NLP-Based-Research-Paper-Intelligence-System\data\processed_data"

os.makedirs(output_folder, exist_ok=True)


# ==========================================
# 3. Required columns
# ==========================================

required_columns = [
    "Title",
    "Author",
    "Institution",
    "Topic",
    "Domain",
    "Field",
    "Concept",
    "Abstract",
    "Keyword",
    "Citation count",
    "Cites",
    "Top 1% cited",
    "Top 10% cited",
    "DOI"
    
]


# ==========================================
# 4. Find CSV files
# ==========================================

csv_files = glob.glob(
    os.path.join(input_folder, "*.csv")
)

print("CSV files found:", len(csv_files))

for file in csv_files:
    print("-", os.path.basename(file))


# ==========================================
# 5. Check whether files were found
# ==========================================

if len(csv_files) == 0:
    raise FileNotFoundError(
        "No CSV files found in the specified folder."
    )


# ==========================================
# 6. Read and process each CSV
# ==========================================

dataframes = []

for file in csv_files:

    print("\nProcessing:", os.path.basename(file))

    df = pd.read_csv(
        file,
        encoding="utf-8-sig"
    )

    print("Rows:", len(df))
    print("Columns:", len(df.columns))

    # Check missing required columns
    missing_columns = [
        column
        for column in required_columns
        if column not in df.columns
    ]

    if missing_columns:

        print("Missing columns:")

        for column in missing_columns:
            print("  -", column)

        # Create missing columns
        for column in missing_columns:
            df[column] = ""


    # Keep only required columns
    df = df[required_columns]

    dataframes.append(df)


# ==========================================
# 7. Merge all CSVs
# ==========================================

merged_df = pd.concat(
    dataframes,
    ignore_index=True
)


# ==========================================
# 8. Remove completely empty rows
# ==========================================

merged_df = merged_df.dropna(
    how="all"
)


# ==========================================
# 9. Remove duplicate titles
# ==========================================

before = len(merged_df)

merged_df = merged_df.drop_duplicates(
    subset=["Title"],
    keep="first"
)

after = len(merged_df)

print("\nDuplicate rows removed:",
      before - after)


# ==========================================
# 10. Ensure correct column order
# ==========================================

merged_df = merged_df[required_columns]


# ==========================================
# 11. Save final dataset
# ==========================================

output_file = os.path.join(
    output_folder,
    "merged_research_papers.csv"
)

merged_df.to_csv(
    output_file,
    index=False,
    encoding="utf-8-sig"
)


# ==========================================
# 12. Final report
# ==========================================

print("\n========================================")
print("MERGING COMPLETED")
print("========================================")

print("Total papers:", len(merged_df))
print("Total columns:", len(merged_df.columns))

print("\nFinal columns:")

for number, column in enumerate(
    merged_df.columns,
    start=1
):
    print(number, ":", column)

print("\nOutput file:")
print(output_file)
```
